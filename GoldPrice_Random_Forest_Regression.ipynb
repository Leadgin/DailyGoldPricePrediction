```python
import pandas as pd #tabular data, data frame
import joblib
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler
import warnings
from sklearn.exceptions import ConvergenceWarning
import time 
import numpy as np
```

```python
Xtrain = pd.read_csv("x_train.csv")
Xtest = pd.read_csv("x_test.csv")
ytrain = pd.read_csv("y_train.csv")
ytest = pd.read_csv("y_test.csv")
```

```python
# Random Forest Regression
from sklearn.ensemble import RandomForestRegressor

# Record starting time
start = time.perf_counter()

rf_regressor = RandomForestRegressor(n_estimators=100, max_depth=10,random_state=42, n_jobs=-1)
rf_regressor.fit(Xtrain, ytrain)

train_accuracy = rf_regressor.score(Xtrain, ytrain)
test_accuracy = rf_regressor.score(Xtest, ytest)

# Record ending time
end = time.perf_counter()

print(f"RF Regressor Train Accuracy (R2): {train_accuracy:.4f}")
print(f"RF Regressor Test Accuracy (R2): {test_accuracy:.8f}")
print(f"Execution time: {end - start:.4f} seconds")

# # Save model to pkl file 
# joblib.dump(rf_regressor, 'rf_regressor_model.pkl')
# print("\nModel saved successfully as rf_regressor_model.pkl!")
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np

# Predictions
ytrain_pred = rf_regressor.predict(Xtrain)
ytest_pred = rf_regressor.predict(Xtest)

# RMSE and MAE for train set
train_rmse = np.sqrt(mean_squared_error(ytrain, ytrain_pred))
train_mae = mean_absolute_error(ytrain, ytrain_pred)

# RMSE and MAE for test set
test_rmse = np.sqrt(mean_squared_error(ytest, ytest_pred))
test_mae = mean_absolute_error(ytest, ytest_pred)

print(f"Train RMSE: {train_rmse:.4f}")
print(f"Train MAE: {train_mae:.4f}")
print(f"Test RMSE: {test_rmse:.4f}")
print(f"Test MAE: {test_mae:.4f}")
```

```python
import time
import joblib
from sklearn.model_selection import GridSearchCV

# Record starting time
start = time.perf_counter()

# Base model
rf = RandomForestRegressor(random_state=42, n_jobs=-1)

# Parameter grid (necessary parameters only)
param_grid = {
    'n_estimators': [100, 200, 500],   # number of trees
    'max_depth': [None, 10, 20],       # tree depth
    'min_samples_split': [2, 5],       # min samples to split
    'min_samples_leaf': [1, 2]         # min samples at leaf
}

# GridSearchCV setup
grid_search_rf2 = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=3,                # safer choice if small dataset
    n_jobs=-1,
    scoring='r2'         # R-squared for regression
)

# Fit grid search
grid_search_rf2.fit(Xtrain, ytrain)

# Record ending time
end = time.perf_counter()

# Evaluate
print(f"Best Parameters: {grid_search_rf2.best_params_}")
print(f"Best CV R2 Score: {grid_search_rf2.best_score_:.4f}")
print(f"Train R2: {grid_search_rf2.best_estimator_.score(Xtrain, ytrain):.4f}")
print(f"Test R2: {grid_search_rf2.best_estimator_.score(Xtest, ytest):.4f}")
print(f"Execution time: {end - start:.4f} seconds")

# Save best model
joblib.dump(grid_search_rf2.best_estimator_, 'rf_regression_best.pkl')
print("\nBest model saved successfully as rf_regression_best.pkl!")
```
