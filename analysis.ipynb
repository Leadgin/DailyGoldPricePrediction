```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid")

# Create a single figure (just for the histogram)
plt.figure(figsize=(10, 6))

# Graph: Histogram with KDE (Kernel Density Estimate) curve
sns.histplot(goldDataset['Price'], kde=True, bins=50, color='gold')

# Add Titles and Labels
plt.title('Figure 3.3.3.2: Distribution of Daily Gold Prices (Histogram)', fontsize=14)
plt.xlabel('Price (INR)', fontsize=12)
plt.ylabel('Frequency (Number of Days)', fontsize=12)

# Display the graph clearly
plt.tight_layout()
plt.show()
```

```python
# boxplot
plt.figure(figsize=(10, 5))

sns.boxplot(
    x=goldDataset["Price"],
    color="gold"
)

plt.title(
    "Figure 3.3.4.1: Spread and Outliers of Gold Prices (Box Plot)",
    fontsize=14
)

plt.xlabel("Price (INR)", fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# 1. Get the 25th percentile (Q1) and 75th percentile (Q3)
Q1 = goldDataset['Price'].quantile(0.25)
Q3 = goldDataset['Price'].quantile(0.75)

# 2. Calculate the IQR
IQR = Q3 - Q1

# 3. Print the result
print(f"Q1: {Q1}")
print(f"Q3: {Q3}")
print(f"The Interquartile Range (IQR) for Price is: {IQR}")
```

```python
# Parse dates (DD-MM-YY format) and sort chronologically
goldDataset['Date'] = pd.to_datetime(goldDataset['Date'], format='%d-%m-%y')
goldDataset = goldDataset.sort_values('Date')

# Plot Scatter Graph
plt.figure(figsize=(10, 5))
plt.scatter(goldDataset['Date'], goldDataset['Price'], color='#DAA520', s=10)

# Formatting
plt.title('Gold Price vs Date', fontsize=14, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Price', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
